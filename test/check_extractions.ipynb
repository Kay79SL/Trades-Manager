```python
"""Quick spot-check of email extractions."""
import json
import os
from dotenv import load_dotenv
from pymongo import MongoClient

load_dotenv()
db = MongoClient(os.getenv("MONGO_URI"))[os.getenv("MONGO_DB", "trades_quotes")]

# Counts
total = db.emails.count_documents({})
extracted = db.emails.count_documents({"extracted": {"$exists": True}})
print(f"Total emails:    {total}")
print(f"With extracted:  {extracted}")
print()

# Show 3 random extractions
print("=" * 60)
print("Sample extractions (3 random):")
print("=" * 60)
for e in db.emails.aggregate([
    {"$match": {"extracted": {"$exists": True}}},
    {"$sample": {"size": 3}},
]):
    print(f"\n--- {e['email_id']} ---")
    print(f"From:    {e.get('from_email', '')}")
    print(f"Subject: {e.get('subject', '')[:80]}")
    print(f"Extracted:")
    extracted = e.get("extracted", {})
    for k, v in extracted.items():
        if k.startswith("_"):
            continue
        print(f"  {k:25} {v}")
```
